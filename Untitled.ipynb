```python
"""
Асинхронное геокодирование большого датафрейма с чекпоинтами.

Вход:
    input.csv с колонками:
        unique_key
        full_address

Выход:
    geocoding_output/geocode_results_part_0001.csv, ...
    geocoding_output/processed_unique_keys.txt
    geocoding_output/failed_unique_keys.txt
    geocoding_results_final.csv — после завершения всех частей
"""

import asyncio
import time
from pathlib import Path

import aiohttp
import pandas as pd

# =========================
# НАСТРОЙКИ
# =========================

INPUT_FILE = "input.csv"
OUTPUT_DIR = Path("geocoding_output")
PROCESSED_KEYS_FILE = OUTPUT_DIR / "processed_unique_keys.txt"
FAILED_KEYS_FILE = OUTPUT_DIR / "failed_unique_keys.txt"
FINAL_RESULT_FILE = Path("geocoding_results_final.csv")

# Каждые 10 000 успешных геокодирований:
# - записывается новый CSV с координатами;
# - сбрасывается буфер обработанных unique_key.
CHUNK_SIZE = 10_000

# Подставьте параметры вашего геокодера.
GEOCODER_URL = "https://api.example.com/geocode"
API_KEY = "YOUR_API_KEY"

# Начните с небольшого значения и повышайте, только если API
# не возвращает 429 и не блокирует запросы.
TARGET_RPS = 45
MAX_CONCURRENT_REQUESTS = 45

REQUEST_TIMEOUT = 20
MAX_RETRIES = 5
RETRY_BASE_DELAY = 1.0

# Адаптируйте под формат ответа вашего API.
# Предполагается ответ вида:
# {
#     "status": "OK",
#     "results": [
#         {"lat": 55.7558, "lon": 37.6173}
#     ]
# }
RESPONSE_RESULTS_FIELD = "results"
RESPONSE_LAT_FIELD = "lat"
RESPONSE_LON_FIELD = "lon"
SUCCESS_STATUS = "OK"

# =========================
# RATE LIMITER
# =========================

class AsyncRateLimiter:
    """
    Token-bucket rate limiter.

    Поддерживает устойчивую частоту запросов TARGET_RPS,
    не превышая разрешенный лимит API.
    """

    def __init__(self, rate_per_second: float):
        self.rate = rate_per_second
        self.capacity = max(1.0, rate_per_second)
        self.tokens = self.capacity
        self.updated_at = time.monotonic()
        self.lock = asyncio.Lock()

    async def acquire(self):
        async with self.lock:
            while True:
                now = time.monotonic()
                elapsed = now - self.updated_at

                self.tokens = min(
                    self.capacity,
                    self.tokens + elapsed * self.rate,
                )
                self.updated_at = now

                if self.tokens >= 1:
                    self.tokens -= 1
                    return

                wait_time = (1 - self.tokens) / self.rate
                await asyncio.sleep(wait_time)

# =========================
# БУФЕР ОБРАБОТАННЫХ КЛЮЧЕЙ
# =========================

class ProcessedKeysBuffer:
    """
    Накапливает успешно геокодированные unique_key
    и записывает их в файл пачками по flush_size.

    При аварийном завершении не сброшенные ключи будут потеряны,
    поэтому соответствующие адреса будут повторно геокодированы
    при следующем запуске. Потери ограничены flush_size записями.
    """

    def __init__(self, path: Path, flush_size: int = 10_000):
        self.path = path
        self.flush_size = flush_size
        self.buffer: list[str] = []

    def add(self, key: str):
        self.buffer.append(key)

        if len(self.buffer) >= self.flush_size:
            self.flush()

    def flush(self):
        if not self.buffer:
            return

        # Режим "a" сохраняет ключи предыдущих запусков.
        with self.path.open("a", encoding="utf-8") as f:
            f.write("\n".join(self.buffer))
            f.write("\n")

        self.buffer.clear()

    def __len__(self):
        return len(self.buffer)

# =========================
# ЗАПРОС К ГЕОКОДЕРУ
# =========================

async def geocode_one(
    session: aiohttp.ClientSession,
    limiter: AsyncRateLimiter,
    semaphore: asyncio.Semaphore,
    unique_key: str,
    address: str,
):
    params = {
        "address": address,
        "key": API_KEY,  # адаптируйте имя параметра под ваш API
        "format": "json",
    }

    delay = RETRY_BASE_DELAY

    for attempt in range(1, MAX_RETRIES + 1):
        try:
            await limiter.acquire()

            async with semaphore:
                async with session.get(
                    GEOCODER_URL,
                    params=params,
                    timeout=aiohttp.ClientTimeout(total=REQUEST_TIMEOUT),
                ) as response:
                    if response.status == 429:
                        retry_after = response.headers.get("Retry-After")

                        wait = (
                            float(retry_after)
                            if retry_after
                            else delay
                        )

                        await asyncio.sleep(wait)
                        delay *= 2
                        continue

                    if response.status >= 500:
                        await asyncio.sleep(delay)
                        delay *= 2
                        continue

                    response.raise_for_status()
                    data = await response.json()

                    results = data.get(RESPONSE_RESULTS_FIELD) or []

                    if data.get("status") != SUCCESS_STATUS or not results:
                        return unique_key, None, None, "not_found"

                    first_result = results[0]

                    return (
                        unique_key,
                        first_result.get(RESPONSE_LAT_FIELD),
                        first_result.get(RESPONSE_LON_FIELD),
                        "ok",
                    )

        except (aiohttp.ClientError, asyncio.TimeoutError) as exc:
            if attempt == MAX_RETRIES:
                return (
                    unique_key,
                    None,
                    None,
                    f"error: {type(exc).__name__}",
                )

            await asyncio.sleep(delay)
            delay *= 2

        except Exception as exc:
            return unique_key, None, None, f"error: {exc}"

    return unique_key, None, None, "max_retries_exceeded"

# =========================
# ЧЕКПОИНТЫ
# =========================

def load_processed_keys(path: Path) -> set[str]:
    if not path.exists():
        return set()

    keys = set()

    with path.open("r", encoding="utf-8") as f:
        for line in f:
            key = line.strip()

            if key:
                keys.add(key)

    return keys

def append_failed_key(path: Path, key: str):
    with path.open("a", encoding="utf-8") as f:
        f.write(f"{key}\n")

def get_next_chunk_number(output_dir: Path) -> int:
    existing_files = list(
        output_dir.glob("geocode_results_part_*.csv")
    )

    return len(existing_files) + 1

# =========================
# ОСНОВНОЙ PIPELINE
# =========================

async def run():
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    print("Чтение входного файла...")

    df = pd.read_csv(
        INPUT_FILE,
        usecols=["unique_key", "full_address"],
        dtype={"unique_key": str, "full_address": str},
    )

    print(f"Всего записей: {len(df):,}")

    processed_keys = load_processed_keys(PROCESSED_KEYS_FILE)
    print(f"Уже геокодировано: {len(processed_keys):,}")

    df = df[~df["unique_key"].isin(processed_keys)].copy()
    df = df.dropna(subset=["full_address"])
    df = df.drop_duplicates(subset=["unique_key"])

    addresses_to_geocode = list(
        df.itertuples(index=False, name=None)
    )

    print(f"Осталось геокодировать: {len(addresses_to_geocode):,}")

    if not addresses_to_geocode:
        print("Новых адресов для геокодирования нет.")
        return

    limiter = AsyncRateLimiter(TARGET_RPS)
    semaphore = asyncio.Semaphore(MAX_CONCURRENT_REQUESTS)

    processed_keys_buffer = ProcessedKeysBuffer(
        path=PROCESSED_KEYS_FILE,
        flush_size=CHUNK_SIZE,
    )

    connector = aiohttp.TCPConnector(
        limit=MAX_CONCURRENT_REQUESTS + 10,
        limit_per_host=MAX_CONCURRENT_REQUESTS,
        ttl_dns_cache=300,
    )

    current_chunk: list[dict] = []
    chunk_number = get_next_chunk_number(OUTPUT_DIR)

    success_count = 0
    failed_count = 0
    started_at = time.monotonic()

    async with aiohttp.ClientSession(connector=connector) as session:
        tasks = [
            geocode_one(
                session=session,
                limiter=limiter,
                semaphore=semaphore,
                unique_key=unique_key,
                address=address,
            )
            for unique_key, address in addresses_to_geocode
        ]

        for i, task in enumerate(
            asyncio.as_completed(tasks),
            start=1,
        ):
            unique_key, latitude, longitude, status = await task

            if status == "ok":
                current_chunk.append(
                    {
                        "unique_key": unique_key,
                        "latitude": latitude,
                        "longitude": longitude,
                        "status": status,
                    }
                )

                processed_keys_buffer.add(unique_key)
                success_count += 1

            else:
                append_failed_key(FAILED_KEYS_FILE, unique_key)
                failed_count += 1

            # Сначала фиксируем ключи, затем пишем CSV.
            # При падении между этими операциями возможно
            # повторное геокодирование не более CHUNK_SIZE адресов.
            if len(current_chunk) >= CHUNK_SIZE:
                processed_keys_buffer.flush()

                output_file = OUTPUT_DIR / (
                    f"geocode_results_part_{chunk_number:04d}.csv"
                )

                pd.DataFrame(current_chunk).to_csv(
                    output_file,
                    index=False,
                    encoding="utf-8-sig",
                )

                print(
                    f"Сохранено {len(current_chunk):,} строк: "
                    f"{output_file}"
                )

                current_chunk = []
                chunk_number += 1

            if i % 10_000 == 0:
                elapsed = time.monotonic() - started_at
                speed = i / elapsed if elapsed else 0
                remaining_seconds = (
                    (len(addresses_to_geocode) - i) / speed
                    if speed
                    else 0
                )

                print(
                    f"Обработано {i:,}/{len(addresses_to_geocode):,} | "
                    f"{speed:.1f} req/s | "
                    f"успешно {success_count:,} | "
                    f"ошибок {failed_count:,} | "
                    f"осталось ~{remaining_seconds / 3600:.1f} ч"
                )

    # Финальный неполный chunk.
    if current_chunk:
        processed_keys_buffer.flush()

        output_file = OUTPUT_DIR / (
            f"geocode_results_part_{chunk_number:04d}.csv"
        )

        pd.DataFrame(current_chunk).to_csv(
            output_file,
            index=False,
            encoding="utf-8-sig",
        )

        print(
            f"Сохранено {len(current_chunk):,} строк: {output_file}"
        )

    # На случай, если после последней записи CSV осталось
    # меньше CHUNK_SIZE успешных ключей.
    processed_keys_buffer.flush()

    print("Геокодирование завершено.")

# =========================
# ФИНАЛЬНАЯ СБОРКА
# =========================

def merge_result_files():
    result_files = sorted(
        OUTPUT_DIR.glob("geocode_results_part_*.csv")
    )

    if not result_files:
        print("Нет файлов с результатами для объединения.")
        return

    print(f"Объединение {len(result_files)} CSV-файлов...")

    frames = []

    for file_path in result_files:
        frame = pd.read_csv(
            file_path,
            dtype={"unique_key": str},
        )
        frames.append(frame)

    result_df = pd.concat(frames, ignore_index=True)

    # Защита от дублей, если программа упала после записи ключей,
    # но до записи соответствующего CSV.
    result_df = result_df.drop_duplicates(
        subset=["unique_key"],
        keep="first",
    )

    result_df.to_csv(
        FINAL_RESULT_FILE,
        index=False,
        encoding="utf-8-sig",
    )

    print(
        f"Итоговый файл сохранен: {FINAL_RESULT_FILE} "
        f"({len(result_df):,} строк)"
    )

if __name__ == "__main__":
    asyncio.run(run())
    merge_result_files()
```
